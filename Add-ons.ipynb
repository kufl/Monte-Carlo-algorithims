## Python Script Add-on
For each Monte Carlo Algorithims with multirun simulations, we sometimes would like to inspect or collect all the individual estimates instead of just obtaining the statistical computation value. 

Hence, a prompt can be added to the existing python script to allow this function.

### A Simple Input Loop
We can utilise a `if-elif-else` chain and `inputimeout` to prompt the user to avoid clutter when the full list of estimates are not needed. Of course, this can be improved upon, especially if it is common that the user will sometimes only wish for a specific amount of estimates only! 

It is highly advisable to have a timeout for user inputs in simulation algorithms to safeguard against infinite loops or any unnecessary workload on server resources. Thus, the standard Python `input` is not the best choice and should be disregarded. Instead, `inputimeout` and `TimeoutOccured` is greatly recommended.

```python
from inputimeout import inputimeout, TimeoutOccurred

def main():
# in the main() function

    estimates = []
    """ estimates are included here to ensure this example script runs without error.
    In practice, you would replace this with your actual estimates. """
    print("Do you want to see the estimates? (y/n)")
    while True:
        try:
            user_input = inputimeout(prompt="", timeout=5).lower()
            # 5 second timer
            if user_input == 'y':
                print(estimates)
                break
            elif user_input == 'n':
                print("Okay, Session is terminated.")
                break
            else:
                print("Invalid input. Session is terminated")
                break

        except TimeoutOccurred:
            print("Timeout. Session is terminated.")
            break

if __name__ == "__main__":
    main()
```
